Task 06

```python
areas = np.array(["Punjab", "Sindh", "KP", "Balochistan", "Islamabad", "AJK"])
pop17 = np.array([110.01, 47.85, 35.53, 12.34, 2.01, 4.05])
pop_2023 = np.array([127.69, 55.70, 40.85, 14.89, 2.36, 4.05])

def as_dict(pairs):
    return {str(k_idx): (elem.item() if isinstance(elem, np.generic) else elem) for k_idx, elem in pairs}

grand_total = pop_2023.sum()
print(f"[T06a] total population 2023 = {grand_total:.2f} million")

share_of_total = pop_2023 / grand_total * 100
print(f"[T06b] share of total (%): {as_dict(zip(areas, np.round(share_of_total, 2)))}")

print(f"[T06c] most populous unit: {areas[np.argmax(pop_2023)]}")
print(f"[T06d] units above 20 million: {areas[np.where(pop_2023 > 20)]}")

growth = ((pop_2023 / pop17) ** (1 / 6) - 1) * 100
print(f"[T06e] annual growth 2017-2023 (% per year): {as_dict(zip(areas, np.round(growth, 2)))}")

print(f"[T06f] fastest-growing unit: {areas[np.argmax(growth)]} ({growth.max():.2f}% per year)")

nat_growth = ((pop_2023.sum() / pop17.sum()) ** (1 / 6) - 1) * 100
double_yrs = 70 / growth.max()
print(f"[T06 interpret] national growth = {nat_growth:.2f}% per year.")
print(f"[T06 interpret] {areas[np.argmax(growth)]} would double in about {double_yrs:.0f} years at its current rate.")
print(f"[T06 interpret] AJK grew {growth[-1]:.2f}% per year vs national {nat_growth:.2f}% -> "
      f"AJK is {nat_growth - growth[-1]:.2f} points below the national rate (AJK is essentially flat in this sample).")
```

Task 07

```python
city_list = np.array(["Lahore", "Karachi", "Islamabad", "Quetta", "Peshawar"])
mon_names = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                   "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])

temp = np.array([
    [19, 22, 28, 35, 40, 40, 36, 34, 34, 32, 26, 21],
    [26, 28, 32, 34, 35, 34, 32, 31, 32, 34, 31, 28],
    [17, 19, 24, 30, 36, 38, 35, 33, 33, 30, 24, 19],
    [11, 13, 18, 24, 29, 34, 35, 34, 31, 25, 19, 13],
    [18, 20, 25, 31, 37, 40, 38, 36, 35, 31, 25, 19],
])
rainfall = np.array([
    [12, 22, 34, 22, 21, 54, 214, 232, 82, 8, 3, 8],
    [8, 10, 9, 4, 1, 6, 40, 25, 13, 1, 2, 5],
    [56, 65, 76, 52, 32, 66, 260, 310, 120, 28, 12, 28],
    [38, 40, 33, 20, 8, 5, 18, 15, 5, 3, 7, 18],
    [33, 48, 63, 48, 23, 21, 86, 95, 43, 23, 20, 29],
])

def as_dict(pairs):
    return {str(k_idx): (elem.item() if isinstance(elem, np.generic) else elem) for k_idx, elem in pairs}

print(f"[T07a] shape of temp (cities, months): {temp.shape}")

mean_yearly = temp.mean(axis=1)
print(f"[T07b] yearly mean temperature (deg C): {as_dict(zip(city_list, np.round(mean_yearly, 1)))}")

hottest_mon = mon_names[temp.argmax(axis=1)]
print(f"[T07c] hottest month per city: {as_dict(zip(city_list, hottest_mon))}")

print(f"[T07d] 5-city monthly average (deg C): {as_dict(zip(mon_names, np.round(temp.mean(axis=0), 1)))}")

count35 = np.sum(temp > 35, axis=1)
print(f"[T07e] months above 35 deg C: {as_dict(zip(city_list, count35))}")

lahore_f = temp[0] * 1.8 + 32
print(f"[T07f] Lahore monthly temperature (deg F): {np.round(lahore_f, 1)}")

monsoon_rain = rainfall[:, 6:9].sum(axis=1)
print(f"[T07g] monsoon rain Jul-Sep (mm): {as_dict(zip(city_list, monsoon_rain))}")
print(f"[T07g] wettest monsoon city: {city_list[np.argmax(monsoon_rain)]} ({monsoon_rain.max()} mm)")

share_of_total = 100 * monsoon_rain / rainfall.sum(axis=1)
print(f"[T07h] monsoon share of annual rain (%): {as_dict(zip(city_list, np.round(share_of_total, 1)))}")

k_idx = list(city_list).index("Karachi")
print(f"[T07 interpret] Karachi has the highest yearly mean ({mean_yearly.max():.1f} deg C) "
      f"but {count35[k_idx]} months above 35 deg C, while cooler-on-average Peshawar has {count35[4]} "
      f"and Lahore {count35[0]}. A high yearly mean does not mean the most extreme heat: "
      f"Karachi is hot all year, the inland cities spike in summer.")
```

Task 08

```python
fy_labels = np.array(["FY2023", "FY2024", "FY2025"])
fmonths = np.array(["Jul", "Aug", "Sep", "Oct", "Nov", "Dec",
                    "Jan", "Feb", "Mar", "Apr", "May", "Jun"])

inflation = np.array([
    [24.9, 27.3, 23.2, 26.6, 23.8, 24.5, 27.6, 31.5, 35.4, 36.4, 38.0, 29.4],
    [28.3, 27.4, 31.4, 26.9, 29.2, 29.7, 28.3, 23.1, 20.7, 17.3, 11.8, 12.6],
    [4.1, 9.6, 6.9, 7.2, 4.9, 4.1, 2.4, 1.5, 0.7, 0.3, 3.5, 3.2],
])

def as_dict(pairs):
    return {str(k_idx): (elem.item() if isinstance(elem, np.generic) else elem) for k_idx, elem in pairs}

mean_infl = inflation.mean(axis=1)
print(f"[T08a] average inflation per year (%): {as_dict(zip(fy_labels, np.round(mean_infl, 2)))}")

infl_fy24 = inflation[1]
print(f"[T08b] FY2024 months above 25 %: {fmonths[infl_fy24 > 25]} -> {np.sum(infl_fy24 > 25)} months")

monthly_change = inflation[:, 1:] - inflation[:, :-1]
print(f"[T08c] month-on-month change (percentage points; negative = drop):\n{np.round(monthly_change, 1)}")

biggest = np.argmin(monthly_change, axis=1)
print(f"[T08d] month of biggest drop: {as_dict(zip(fy_labels, fmonths[biggest + 1]))} | size (pp): {np.round(monthly_change.min(axis=1), 1)}")

infl25 = inflation[2]
level = np.where(infl25 < 5, "Low", np.where(infl25 < 10, "Moderate", "High"))
print(f"[T08e] FY2025 labels (Low < 5 %, Moderate 5-10 %, High >= 10 %):\n{np.column_stack((fmonths, infl25, level))}")

print(f"[T08 interpret] average inflation went {mean_infl[0]:.1f}% -> {mean_infl[1]:.1f}% -> {mean_infl[2]:.1f}%. "
      f"Prices did NOT fall; they kept rising, but more and more slowly (disinflation), "
      f"so the price LEVEL is still much higher than at the start of FY2023.")
```

Task 09

```python
marks_arr = np.array([34, 38, 35, 52, 67, 73, 88, 95, 98, 39,
                  41, 45, 36, 58, 62, 71, 80, 33, 99, 44])
roll_tags = np.arange(1, 21)
PASS = 40

marks_after = np.clip(marks_arr + 5, 0, 100)
print(f"[T09a] original marks: {marks_arr}")
print(f"[T09a] marks after +5 grace (cap 100): {marks_after}")

pct_before = np.mean(marks_arr >= PASS) * 100
p_after = np.mean(marks_after >= PASS) * 100
print(f"[T09b] pass % before grace = {pct_before:.1f} %, after grace = {p_after:.1f} %")

just_grace = np.logical_and(marks_arr < PASS, marks_after >= PASS)
print(f"[T09c] passed only because of grace (roll numbers): {roll_tags[just_grace]}")
print(f"[T09c] their original marks: {marks_arr[just_grace]}")

z = (marks_after - marks_after.mean()) / marks_after.std()
print(f"[T09d] mean = {marks_after.mean():.2f}, std = {marks_after.std():.2f}")
print(f"[T09d] z-scores (std devs from mean): {np.round(z, 2)}")
```

Task 10

```python
img = np.zeros((8, 8), dtype=np.uint8)
img[1, 1:7] = 255
for r, c in zip([2, 3, 4, 5, 6], [6, 5, 5, 4, 4]):
    img[r, c] = 255

as_bits = lambda a: (a > 0).astype(int)

print(f"[T10a] the '7' (1 = white):\n{as_bits(img)}")

flip_lr = img[:, ::-1]
print(f"[T10b] flipped left-right:\n{as_bits(flip_lr)}")

invert_img = 255 - img
print(f"[T10c] inverted (white <-> black):\n{as_bits(invert_img)}")

cropped = img[:7, 1:]
print(f"[T10d] cropped 7x7 region, shape: {cropped.shape}\n{as_bits(cropped)}")

brightened = np.clip(img.astype(np.int16) + 60, 0, 255).astype(np.uint8)
print(f"[T10e] brightened (+60, capped at 255); unique values: {np.unique(brightened)}")

print(f"[T10f] shape = {img.shape} | dtype = {img.dtype} | white pixels = {np.sum(img == 255)}")
```
